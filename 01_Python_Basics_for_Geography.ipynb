## 1. ตัวแปรและชนิดข้อมูล (Variables & Data types)

```python
# ตัวแปร = กล่องเก็บค่า  ชื่อตัวแปร = ค่า
station_name = "เชียงใหม่"     # str   (ข้อความ) ต้องอยู่ในเครื่องหมายคำพูด
latitude     = 18.79           # float (ทศนิยม) ละติจูด หน่วยองศา
longitude    = 98.98           # float ลองจิจูด
elevation_m  = 314             # int   (จำนวนเต็ม) ความสูงจากระดับทะเล (เมตร)
is_active    = True            # bool  (จริง/เท็จ) สถานียังเปิดใช้งานหรือไม่

print(station_name, latitude, longitude, elevation_m, is_active)
print(type(station_name), type(latitude), type(elevation_m), type(is_active))
```

```python
# f-string: ใส่ตัวแปรลงในข้อความ ใช้ f"...{ตัวแปร}..."
# :.2f หมายถึงแสดงทศนิยม 2 ตำแหน่ง
print(f"สถานี{station_name} ตั้งอยู่ที่ ({latitude:.2f}°N, {longitude:.2f}°E) สูง {elevation_m} ม.")
```

```python
# การคำนวณพื้นฐาน: อุณหภูมิลดลงตามความสูง (Environmental lapse rate ≈ 6.5 °C ต่อ 1 กม.)
temp_at_sea_level = 32.0                    # °C
lapse_rate = 6.5 / 1000                     # °C ต่อเมตร
temp_station = temp_at_sea_level - lapse_rate * elevation_m
print(f"อุณหภูมิโดยประมาณที่สถานี = {temp_station:.1f} °C")

# ลองที่ดอยอินทนนท์ (สูง ~2,565 ม.)
print(f"อุณหภูมิโดยประมาณที่ยอดดอยอินทนนท์ = {temp_at_sea_level - lapse_rate*2565:.1f} °C")
```

## 2. List: เก็บข้อมูลหลายค่าเรียงกัน

```python
# ปริมาณฝนรายเดือน (มม.) ของสถานีตัวอย่าง ม.ค. → ธ.ค.
months = ["ม.ค.","ก.พ.","มี.ค.","เม.ย.","พ.ค.","มิ.ย.","ก.ค.","ส.ค.","ก.ย.","ต.ค.","พ.ย.","ธ.ค."]
rain   = [5.2, 8.1, 20.4, 55.0, 160.3, 140.8, 180.2, 235.6, 210.4, 105.7, 30.2, 7.5]

print("จำนวนเดือน  :", len(rain))
print("ฝนเดือนแรก  :", rain[0])       # Python นับตำแหน่ง (index) เริ่มจาก 0
print("ฝนเดือนสุดท้าย:", rain[-1])     # -1 คือตัวสุดท้าย
print("ฝน มิ.ย.–ส.ค.:", rain[5:8])     # slicing: ตั้งแต่ index 5 ถึง 7 (ไม่รวม 8)
```

```python
total = sum(rain)                 # ฝนรวมทั้งปี
mean  = total / len(rain)         # ค่าเฉลี่ยรายเดือน
wettest_index = rain.index(max(rain))   # ตำแหน่งของค่ามากที่สุด

print(f"ฝนรวมรายปี     = {total:.1f} มม.")
print(f"ฝนเฉลี่ยรายเดือน = {mean:.1f} มม.")
print(f"เดือนที่ฝนมากที่สุดคือ {months[wettest_index]} ({max(rain)} มม.)")
```

## 3. Dictionary: เก็บข้อมูลแบบ key → value (เหมือน 1 แถวในตาราง attribute ของ GIS)

```python
station = {
    "id": "S01",
    "name": "เชียงใหม่",
    "lat": 18.79,
    "lon": 98.98,
    "elev_m": 314,
    "rain_mm": rain,          # value เป็น list ก็ได้
}
print(station["name"], "→ ความสูง", station["elev_m"], "ม.")
station["province"] = "เชียงใหม่"       # เพิ่ม key ใหม่
print(station.keys())
```

## 4. Loop: ทำซ้ำ

```python
# zip() จับคู่ list สองชุดเข้าด้วยกันทีละตัว
for m, r in zip(months, rain):
    bar = "█" * int(r // 10)          # วาดกราฟแท่งแบบข้อความ 1 ช่อง = 10 มม.
    print(f"{m:>6} {r:6.1f} {bar}")
```

## 5. If–else: จำแนกความรุนแรงของฝน
เกณฑ์ปริมาณฝน **รายวัน** ของกรมอุตุนิยมวิทยา:

| ปริมาณฝน 24 ชม. (มม.) | ระดับ |
|---|---|
| 0.1 – 10.0 | ฝนเล็กน้อย |
| 10.1 – 35.0 | ฝนปานกลาง |
| 35.1 – 90.0 | ฝนหนัก |
| > 90.0 | ฝนหนักมาก |

```python
def classify_daily_rain(mm):
    """จำแนกความรุนแรงฝนรายวันตามเกณฑ์กรมอุตุนิยมวิทยา"""
    if mm < 0.1:
        return "ไม่มีฝน"
    elif mm <= 10.0:
        return "ฝนเล็กน้อย"
    elif mm <= 35.0:
        return "ฝนปานกลาง"
    elif mm <= 90.0:
        return "ฝนหนัก"
    else:
        return "ฝนหนักมาก"

daily_rain = [0, 3.2, 12.5, 48.0, 120.4, 0.05]
for d in daily_rain:
    print(f"{d:6.1f} มม. → {classify_daily_rain(d)}")
```

## 6. Function: รวมขั้นตอนไว้เรียกใช้ซ้ำ
### 6.1 แปลงพิกัด องศา-ลิปดา-ฟิลิปดา (DMS) เป็นองศาทศนิยม (DD)
$$DD = D + \frac{M}{60} + \frac{S}{3600}$$

```python
def dms_to_dd(deg, minute, second, hemisphere="N"):
    """แปลง DMS → Decimal Degrees  ซีกโลกใต้ (S) และตะวันตก (W) ให้ค่าติดลบ"""
    dd = deg + minute/60 + second/3600
    if hemisphere in ("S", "W"):
        dd = -dd
    return dd

# พิกัดสถานีอุตุนิยมวิทยาเชียงใหม่โดยประมาณ 18°47'24"N 98°58'48"E
lat = dms_to_dd(18, 47, 24, "N")
lon = dms_to_dd(98, 58, 48, "E")
print(f"lat = {lat:.5f}, lon = {lon:.5f}")
```

### 6.2 🟡 ระยะทางระหว่างสองจุดบนผิวโลก (Haversine formula)
$$d = 2R\arcsin\sqrt{\sin^2\frac{\Delta\phi}{2} + \cos\phi_1\cos\phi_2\sin^2\frac{\Delta\lambda}{2}}$$
โดย $R$ = รัศมีโลก ≈ 6,371 กม., $\phi$ = ละติจูด, $\lambda$ = ลองจิจูด (หน่วยเรเดียน)

```python
import math

def haversine_km(lat1, lon1, lat2, lon2):
    """ระยะทางวงกลมใหญ่ (great-circle distance) หน่วยกิโลเมตร"""
    R = 6371.0
    phi1, phi2 = math.radians(lat1), math.radians(lat2)     # องศา → เรเดียน
    dphi = math.radians(lat2 - lat1)
    dlmb = math.radians(lon2 - lon1)
    a = math.sin(dphi/2)**2 + math.cos(phi1)*math.cos(phi2)*math.sin(dlmb/2)**2
    return 2 * R * math.asin(math.sqrt(a))

d = haversine_km(18.79, 98.98, 18.78, 100.78)     # เชียงใหม่ → น่าน
print(f"ระยะทางเชียงใหม่–น่าน (เส้นตรง) ≈ {d:.1f} กม.")
```

```python
# ใช้ loop + dictionary + function ร่วมกัน: หาสถานีที่ใกล้จุดสนใจที่สุด
stations = {
    "เชียงใหม่": (18.79, 98.98),
    "ลำปาง":    (18.29, 99.49),
    "น่าน":      (18.78, 100.78),
    "เชียงราย":  (19.91, 99.83),
    "แพร่":      (18.14, 100.14),
}
poi = (18.60, 100.20)     # จุดสนใจ (Point of Interest) เช่น หมู่บ้านที่เกิดดินถล่ม

distances = {}
for name, (la, lo) in stations.items():
    distances[name] = haversine_km(poi[0], poi[1], la, lo)

for name, dist in sorted(distances.items(), key=lambda x: x[1]):   # เรียงจากใกล้ไปไกล
    print(f"{name:10s} {dist:7.1f} กม.")
```
